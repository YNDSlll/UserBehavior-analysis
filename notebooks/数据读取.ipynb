```python
# import pandas as pd

# # header=None：这份数据没有表头行，不写这个参数，
# # 第一行真实数据会被当成列名吃掉（去核对我上次的提示）
# # names=...：手动给 5 列起名。官方字段说明就是这个顺序
# df = pd.read_csv(
#     '../data/raw/UserBehavior.csv',
#     header=None,#文件中没有列名
#     names=['user_id', 'item_id', 'category_id', 'behavior_type', 'timestamp'],
#     nrows=1000000   # 只读前 100 万行。全量 3.4GB 一个亿条，读一次几分钟还吃内存；
#                     # 学习项目用采样完全够，面试就说"基于 100 万条采样"
#     }               
# )

# print(df.shape)   # 期望输出 (1000000, 5)
# df.head()
import pandas as pd

df = pd.read_csv(
    '../data/raw/UserBehavior.csv',
    header=None,
    names=['user_id', 'item_id', 'category_id', 'behavior_type', 'timestamp'],
    nrows=1000000,
    dtype={
        "user_id": "string",
        "item_id": "string",
        "category_id": "string"
    }
)

df.head()
```

```python
print(df.dtypes) ;     # 看每列类型
print(df.isnull().sum());   # 看有没有缺失
print(df.describe() ) ;     # 看数值分布概况
#只用df.describe()会默认只看数值列，其他列不看。要看所有列的概况，得加参数 include='all'。 
# 多个df. 在一起时，运行输出只会显示最后一行，用print全部打印出来
```

```python
df['behavior_type'].unique()  #unique()函数可以返回Series中唯一的值，这里返回的是behavior_type列去重后的所有值，即四种行为类型
```

```python
df['behavior_type'].value_counts()    # 看每种行为各多少条 —— pv 远大于 buy 就是正常电商形态
```

```python
df.to_csv('../data/sample/UserBehavior_1m.csv', index=False)
# index=False：不把行号存成一列，否则下次读进来多一列无意义的 Unnamed: 0
```
