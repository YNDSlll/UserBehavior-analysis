1.时间戳 → 可读时间

```python
#1.时间戳 → 可读时间
import pandas as pd
df = pd.read_csv('../data/sample/UserBehavior_1m.csv', dtype={'user_id': str, 'item_id': str, 'category_id': str})

df['datetime'] = pd.to_datetime(df['timestamp'], unit='s') + pd.Timedelta(hours=8)
df['date'] = df['datetime'].dt.date   # 单独存一列日期，后面按天统计全靠它
df.head(5)
```

```python
#2.按天统计行为数量
daily = df.groupby('date').size()
print(daily)
```

```python
#3.剔除脏数据
start = 1511539200   # 2017-11-25 00:00 北京时间
before = len(df)
df_clean = df[df['timestamp'] >= start].copy()
print(f'剔除脏数据 {before - len(df_clean)} 行，剩余 {len(df_clean)} 行')
```

```python
#4.剔除重复数据+基础盘点
print('完全重复行：', df_clean.duplicated().sum())
df_clean = df_clean.drop_duplicates()
print(df_clean['user_id'].nunique(), '个用户')
print(df_clean['item_id'].nunique(), '个商品')
print(df_clean['category_id'].nunique(), '个类目')
```

```python
#5.保存清洗后的数据
df_clean.to_csv('../data/processed/UserBehavior_clean.csv', index=False)
print(df_clean.head())
```
